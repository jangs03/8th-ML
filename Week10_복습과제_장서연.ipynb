# 1. PCA(Principal Component Analysis)

- PCA 실습은 와인 데이터셋 (Wine)으로 진행합니다. 아래의 셀을 로드하고 실습을 시작해주세요.

```python
from sklearn.datasets import load_wine
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

# 사이킷런 내장 데이터 셋 API 호출
wine = load_wine()

# 넘파이 데이터 셋을 Pandas DataFrame으로 변환
columns = wine.feature_names
wineDF = pd.DataFrame(wine.data, columns=columns)
wineDF['target'] = wine.target
wineDF.head(3)
```

# 1-1. 원본 와인 데이터 세트가 어떻게 분포돼 있는지 두 개의 속성(alcohol,	malic_acid)을 각각 X축, Y축으로 해서 2차원으로 시각화 해주세요.

- ##답안## 을 채워주세요

```python
# wine0는 세모, wine1는 네모, wine2는 동그라미로 표현
markers=['^', 's', 'o']
wine_ind=[]
# wine0의 target 값은 0, wine1는 1, wine2는 2. 각 target별로 다른 모양으로 산점도로 표시
for i, marker in enumerate(markers):
    indices =  wineDF[wineDF['target']==i].index
    wine_ind.append(indices)
    plt.scatter(x= wineDF.loc[wine_ind[i],'alcohol'],
                y= wineDF.loc[wine_ind[i],'malic_acid'],marker=marker,label=f'wine{i}')
plt.legend()
plt.xlabel('alcohol')
plt.ylabel('malic_acid')
plt.show()
```

- PCA 적용 전, 개별 속성 스케일링하는 셀을 실행해주세요

```python
from sklearn.preprocessing import StandardScaler

# Target 값을 제외한 모든 속성 값을 StandardScaler를 이용해 표준 정규 분포를 가지는 값들로 변환
wine_scaled = StandardScaler().fit_transform(wineDF.iloc[:, :-1])
```

# 1-2. PCA 적용 -> 4차원 와인 데이터를 2차원 PCA 데이터로 변환

- ##답안##을 채워주세요.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

# fit()과 transform()을 호출해 PCA 변환 데이터 반환
pca.fit(wine_scaled)
wine_pca = pca.transform(wine_scaled)
print(wine_pca)
```

변환된 PCA 데이터 세트를 DataFrame으로 변환하여 데이터값 확인하는 코드 셀을 실행해주세요.

```python
# PCA 변환된 데이터의 칼럼명을 각각 pca_component_1, pca_component_2로 명명

pca_columns=['pca_component_1','pca_component_2']
wineDF_pca = pd.DataFrame(wine_pca, columns=pca_columns)
wineDF_pca['target']=wine.target
wineDF_pca.head(3)
```

PCA 변환된 데이터 세트 2차원 시각화하는 코드 셀을 실행해주세요.

```python
# wine0를 세모, wine1를 네모, wine2를 동그라미로 표시
markers=['^', 's', 'o']

# pca_component_1 을 x축, pc_component_2를 y축으로 scatter plot 수행
for i, marker in enumerate(markers):
    x_axis_data = wineDF_pca[wineDF_pca['target']==i]['pca_component_1']
    y_axis_data = wineDF_pca[wineDF_pca['target']==i]['pca_component_2']
    plt.scatter(x_axis_data, y_axis_data, marker=marker,label=wine.target_names[i])

plt.legend()
plt.xlabel('pca_component_1')
plt.ylabel('pca_component_2')
plt.show()
```

# 3. SVD

# 3-1) SVD 연산 및 분해하기

랜덤한 2차원 행렬 A를 생성해 특이값 분해하고, 분해된 행렬을 이용해 원본 행렬로 다시 복원해 봅시다.

(1) 행렬 A 생성

: 넘파이의 rand() 함수를 이용해 3x3 크기의 랜덤한 값을 갖는 행렬 A를 생성해 주세요.

```python
import numpy as np

# 3x3 랜덤 행렬 A 생성
np.random.seed(121)
A = np.random.randn(3,3)
print(np.round(A,3))
```

(2) SVD (특이값 분해)

: 넘파이의 SVD 모듈로 A 행렬에 SVD를 적용해 U, Sigma, Vt를 반환받아 그 크기와 값을 출력해 주세요.

(행렬 원소 값은 소수점 아래 3자리까지 반올림해서 출력)

```python
from numpy.linalg import svd

U, Sigma, Vt = svd(A)

# U, Sigma, Vt의 크기 출력
print(U.shape, Sigma.shape, Vt.shape)

# U, Sigma, Vt 행렬 출력
print('U matrix:\n', np.round(U,3))
print('Sigma Value:\n', np.round(Sigma,3))
print('V transpose matrix:\n', np.round(Vt,3))
```

(3) U, Sigma, Vt를 이용해 원본 행렬 A 복원하기

: 넘파이(np)의 dot 연산을 이용해 원본 행렬 A를 계산해 주세요.

(💡 Sigma 행렬은 현재 대각원소 값만을 가지는 1차원 벡터이므로, 0을 포함한 대칭행렬로 변환 후 계산해야 합니다 - 넘파이의 diag 함수 사용!)

```python
# Sigma를 다시 0을 포함한 대칭 행렬로 변환
Sigma_mat = np.diag(Sigma)

# 복원 행렬 A_ 계산
A_ = np.dot(np.dot(U,Sigma_mat),Vt)

print(np.round(A_, 3))
```

# 3-2) row 간 의존성을 갖는 데이터 세트에 SVD 적용하고 복원하기

이번에는 row 간 의존성이 있는 데이터 세트에 SVD를 적용해 Sigma 값이 어떻게 달라지는지 관찰하고, 0이 아닌 Sigma 값만을 이용해 행렬 A를 복원해 봅시다.

(1) A를 row 간 의존성이 있는 값으로 수정하기

: A의 1번째 로우가 2번째 로우와 3번째 로우 값의 합을 가지도록 값을 수정해 주세요.

```python
A[0] = A[1]+A[2]
print(np.round(A, 3))
```

(2) 수정된 A에 SVD를 다시 적용해 U, Sigma, Vt 행렬의 크기와 값을 확인하기

```python
U, Sigma, Vt = svd(A)

# U, Sigma, Vt의 크기 출력
print(U.shape, Sigma.shape, Vt.shape)

# U, Sigma, Vt 행렬 출력
print('U matrix:\n', np.round(U,3))
print('Sigma Value:\n', np.round(Sigma,3))
print('V transpose matrix:\n', np.round(Vt,3))
```

(3) Sigma의 값이 0인 요소를 제외하고 A 복원하기

: Sigma가 0인 데이터를 제외하고 복원해 봅시다. Sigma에서 0 값을 제외했다면, 이에 대응되는 U, Vt의 값도 제외해 주어야 합니다.

```python
# U 행렬의 경우 Sigma와 내적 수행 -> Sigma 데이터가 0이 아닌 값에 대응되는 열만 추출
U_ = U[:,:2]
Sigma_ = Sigma[:2]

# Vt 행렬 또한 Sigma와 내적 수행 -> Sigma 데이터가 0이 아닌 값에 대응되는 행만 추출
Vt_ = Vt[:,:2]

print(U_.shape, Sigma_.shape, Vt_.shape)

# U, Sigma, Vt의 내적을 수행하며, 다시 원본 행렬 복원
A_ = np.dot(np.dot(U_,Sigma_),Vt_)
print(np.round(A_,3))
```

# 5-2. T-SNE를 적용한 데이터를 이용해 시각화를 진행하세요.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
import matplotlib.patches as mpatches

# 시각화 함수 정의
def plot_labelled_scatter(X, y, class_labels, s):
    num_labels = len(class_labels)

    x_min, x_max = X[:, 0].min() -1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() -1, X[:, 1].max() + 1

    color_array = ['#FF6F61', '#6B5B95', '#88B04B', '#F7CAC9', '#92A8D1', '#955251', '#B565A7', '#009B77', '#DD4124', '#45B8AC']
    cmap_bold = ListedColormap(color_array)
    bnorm = BoundaryNorm(np.arange(0, num_labels + 1, 1), ncolors=num_labels)
    plt.figure(figsize=s)

    plt.scatter(X[:, 0], X[:, 1], s=80, c=y, cmap=cmap_bold, norm=bnorm, alpha=0.4, edgecolor='black', lw=1)
    sp = plt.gca().spines
    sp['top'].set_visible(False)
    sp['right'].set_visible(False)

    plt.grid(which='both', color='lightslategrey', alpha=0.3)
    plt.xlim(x_min, x_max)
    plt.ylim(y_min, y_max)

    h = []
    for c in range(0, num_labels):
        h.append(mpatches.Patch(color=color_array[c], label=class_labels[c]))
    plt.legend(handles=h, fontsize=12, frameon=False)

# 시각화 함수 실행
plot_labelled_scatter(
    X_tsne,
    y,
    ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
     'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'],
    (15, 9)
)

plt.xlabel('First T-SNE Dimension', fontsize=14)
plt.ylabel('Second T-SNE Dimension', fontsize=14)
plt.title('Fashion MNIST Dataset T-SNE', fontsize=17)
```

# Dimensionality Reduction for Beginners 캐글 실습

## Loading Depencies and Data

```python
# Dependencies
%matplotlib inline

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
import matplotlib.patches as mpatches
import seaborn as sns
from sklearn.datasets import load_breast_cancer
# Data
cancer = load_breast_cancer()
(X_canc, y_canc) = load_breast_cancer(return_X_y = True)
```

## Helper Functions
: Labelled Scatter Plot 그려주는 함수

```python
def plot_labelled_scatter(X, y, class_labels,s):
    num_labels = len(class_labels)

    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

    marker_array = ['o', '^', '*']
    color_array = ['#FFFF00', '#00AAFF', '#000000', '#FF00AA']
    cmap_bold = ListedColormap(color_array)
    bnorm = BoundaryNorm(np.arange(0, num_labels + 1, 1), ncolors=num_labels)
    plt.figure(figsize=s)

    plt.scatter(X[:, 0], X[:, 1], s=80, c=y, cmap=cmap_bold, norm = bnorm, alpha = 0.4, edgecolor='black', lw = 1)
    sp = plt.gca().spines
    sp['top'].set_visible(False)
    sp['right'].set_visible(False)

    plt.grid(which='both',color='lightslategrey',alpha=0.3)

    plt.xlim(x_min, x_max)
    plt.ylim(y_min, y_max)

    h = []
    for c in range(0, num_labels):
        h.append(mpatches.Patch(color=color_array[c], label=class_labels[c]))
    plt.legend(handles=h,fontsize=15,frameon=False)
```

## 1. PCA ( Principle Component Analysis )
어떤 실수 공간
𝑝-차원의 점 집합에 대해, **주성분(principal components)**은 일련의 방향 벡터로 정의됩니다.
이 벡터들은 데이터에 가장 잘 맞는 직선의 방향을 나타내며, **이전의 주성분 벡터들과는 서로 직교(수직)**합니다.

예를 들어, 우리가 두 개의 원래 특성
$X_1$과 $X_2$ 를 가지고 있고 이 둘이 강하게 상관되어 있다면, 이 점들의 분포는 **타원형 구름(cloud of points)**으로 표현될 수 있습니다.

PCA는 이 데이터를 회전시켜서, 가장 큰 분산 방향을 찾습니다. 이 방향을 **첫 번째 주성분(PC1)**이라고 하며, 구름의 긴 방향을 따라 위치합니다.

그 다음으로, PCA는 이와 수직인 방향에서 남아 있는 분산을 최대한 설명하는 축을 찾습니다. 이 방향이 **두 번째 주성분(PC2)**입니다.

2차원 공간에서는 첫 번째 주성분에 직각인 방향이 오직 하나뿐이지만,
3차원 이상의 고차원 공간에서는 이전 주성분에 직각인 방향이 무수히 많기 때문에, **PCA는 이전 주성분들과 모두 직교하는 방향을 순차적으로 찾아
원하는 수의 주성분이 될 때까지 과정을 반복**합니다.

### Using PCA to find the first two Principal Components of the Breast Cancer Dataset

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Scaling before applying PCA
## zero-centered, unit variance여야 함 >> sklearn의 StandardScaler가 해줌
canc_norm = StandardScaler().fit(X_canc).transform(X_canc)

pca = PCA(n_components=2).fit(canc_norm)
canc_pca = pca.transform(canc_norm)
print('PCA 전 features의 수: {}\nPCA 후 features의 수: {}'.format(X_canc.shape[1],canc_pca.shape[1]))
```

### Plotting the PCA-Transformed Version of the Breast Cancer Dataset

```python
plot_labelled_scatter(canc_pca,y_canc,['malignant','benign'],(15,9))
plt.xlabel('First PC',fontsize = 15)
plt.ylabel('Second PC',fontsize = 15)
plt.title('Breast Cancer Dataset PCA(n_components=2)',fontsize=20);
```

### Using PCA to get feature Co-Relation between the Initial 30 Features

pca의 .components_속성

- shape이 (n_components,n_features)인 배열
- 각 주성분이 원본특성과 얼마나 관련 있는지 나타냄.
- fit하는 동안 생성된 고차원 원본 특성과 저차원 주성분(PC 방향) 사이 co-relation 저장, heatmap 시각화 가능

```python
fig = plt.figure(figsize=(20,9))
plt.imshow(pca.components_,interpolation=None, cmap='viridis')
feature_names = list(cancer.feature_names)

plt.gca().set_xticks(np.arange(len(feature_names)));
plt.gca().set_yticks(np.arange(2));
plt.gca().set_xticklabels(feature_names,rotation=90,fontsize=14);
plt.gca().set_yticklabels(['First PC','Second PC'],fontsize=14);

plt.colorbar(orientation='horizontal',ticks=[pca.components_.min(),0,
                                            pca.components_.max()],pad=0.5);

```

- 1st PC : 모두 (+)값
  * 모든 30개의 특성 간 양의 상관관계 있음
- 2nd PC : (+) & (-) 값 섞임
  * 음수부호를 가진 특성들이 하나의 군집을 형성하며 변동,
  * 나머지 양수 분호들의 특성과는 반대방향으로 변동
  * 예} (mean texture, worst texture ), (mean,radius,worst_radius) 같이 변동함

➡ PCA는 초기에 EDA에 도움이 되고 거의 모든 타입의 데이터셋에 쓸만하지만, 더 복잡한 데이터셋에서 더 나은 세부적인 군집들을 찾기는 힘들다.

## 2. Multi-Dimensional Scaling (MDS)

- Manifold Learning : 고차원 데이터 속에서 저차원 구조를 효과적으로 찾아내는 비지도 학습 알고리즘. 데이터 시각화에 유용
  * manifold - 고차원 공간안에 존재하는 저차원의 곡면 구조
  * 예시) Swiss Roll dataset - 3차원 공간에 데이터가 존재하지만 실제로는 특이한 모양의 2차원 곡면에 모든 점들이 놓여 있음.

- PCA는 복잡/비선형적 구조 찾기 힘듦 -> MDS
- MDS의 종류 多, 공통된 목적은 '고차원 데이터셋 -> 저차원으로 투영, 시각화' + '점들 사이 거리관계(유사도)' 보존

### Using MDS on the Breast Cancer Dataset

```python
from sklearn.manifold import MDS

mds = MDS(n_components=2,random_state=2)
canc_mds = mds.fit_transform(canc_norm)

print('MDS 전 features 수: {}\nMDS 후 features 수: {}'.format(X_canc.shape[1],canc_mds.shape[1]))

plot_labelled_scatter(canc_mds,y_canc,['malignant','benign'],(15,9))
plt.xlabel('First MDS dimension',fontsize=15)
plt.ylabel('Second MDS dimension',fontsize=15)
plt.title('Breast Cancer Dataset MDS(n_components=2)',fontsize=20);
```

## 3. T-distributed Stochastic Neighbor Embedding(T-SNE)
* 시각화에 강력한 manifold learning의 한 종류
* 데이터를 2차원으로 변환, 산점도 상의 점들 간 거리가 고차원 데이터사이에서의 거리관계를 최대한 잘 반영하도록 함.
* 이웃한 점들 사이의 거리관계 중점적으로 보존 ➡ 가까운 데이터 포인트들 가깝게, 멀리 떨어진 점들 관계 덜 중요하게.


### Using T-SNE on the Breast Cancer Dataset  

```python
from sklearn.manifold import TSNE

tsne = TSNE(random_state = 42)

canc_tsne = tsne.fit_transform(canc_norm)

print('T-SNE 전 features 수: {}\nT-SNE 후 features 수: {}'
      .format(X_canc.shape[1],canc_tsne.shape[1]))

plot_labelled_scatter(canc_tsne, y_canc, ['malignant', 'benign'],(15,9))

plt.xlabel('First T-SNE Dimension',fontsize=14)
plt.ylabel('Second T-SNE Dimension',fontsize=14)
plt.title('Breast Cancer Dataset T-SNE',fontsize=17);
```
